```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/creditcard.csv")
df.head()
```

```python
df['Class'].value_counts()
```

```python
counts = pd.Series(df['Class'].values).value_counts()
counts.plot(kind='pie', autopct='%1.1f%%')
plt.title("Class distribution")
plt.show()
```

```python
# Корреляция топ признаков с Class
corr = df.corr()['Class'].sort_values().drop('Class')
corr.plot(kind='barh', figsize=(8, 10), title='Feature correlation with Class')
plt.show()
```
